```python
# ============================================================
# PHASE 2 (var2): Subterranean Thermal Reservoir Mapping
# Ridge vs Lasso vs ElasticNet with POLYNOMIAL FEATURES + GridSearchCV
# compared across several DATA SPLITS (90-10, 85-15, 80-20, 75-25, 70-30).
#
# Inputs : x1, x2, x3  (coordinate offsets)   ->  target y (Thermal Anomaly Score)
# The heat map is a polynomial of degree up to 20, so degrees 1..20 are searched.
#
# For EACH split the TRAIN file is cut into:
#     dev  (GridSearchCV does K-fold CV inside it)  +  internal test (used once at the end)
# The external TEST file (no labels) is only used for the final predictions.
#
# GridSearchCV tunes degree + alpha (+ l1_ratio for ElasticNet) together.
# Winner of each split = lowest CV RMSE. Best split overall = lowest internal-test RMSE
# among those split winners. The best (split, model) is then refit on ALL training rows.
# ============================================================
import os, time, shutil, tempfile, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from joblib import Memory
from sklearn.base import clone
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split, GridSearchCV, KFold
from sklearn.metrics import mean_absolute_error, r2_score

try:
    display  # defined automatically in Colab / Jupyter
except NameError:
    def display(obj):
        print(obj)

warnings.filterwarnings("ignore")
pd.set_option("display.width", 220)
pd.set_option("display.max_columns", 30)
SEED = 42
np.random.seed(SEED)

# ------------------------------------------------------------
# 0. SETTINGS YOU CAN CHANGE  (everything is in this one block)
# ------------------------------------------------------------
TRAIN_FILE = "IMT2024091_train_var2.csv"
TEST_FILE  = "IMT2024091_test_var2.csv"
OUT_FILE   = "IMT2024091_pred_var2.csv"
FEATURES   = ["x1", "x2", "x3"]
TARGET     = "y"

# Internal-test fraction for each split to try.  0.15 means an 85-15 split.
# Remove entries to run faster (each split takes several minutes).
TEST_FRACTIONS = [0.10, 0.15, 0.20, 0.25, 0.30]     # 90-10, 85-15, 80-20, 75-25, 70-30

DEGREES = list(range(1, 21))     # polynomial degrees to try (the task says up to 20)
N_FOLDS = 5                      # K in K-fold cross-validation
N_JOBS  = -1                     # -1 = use all CPU cores

# Hyper-parameter grids
RIDGE_ALPHAS = np.logspace(-3, 3, 13)
LASSO_ALPHAS = np.logspace(-4, 0, 7)
ENET_ALPHAS  = np.logspace(-4, 0, 7)
ENET_L1      = [0.2, 0.5, 0.8]
MAX_ITER, TOL = 20000, 1e-3      # coordinate-descent settings for Lasso / ElasticNet

KINDS = ["ridge", "lasso", "elasticnet"]

# ------------------------------------------------------------
# 1. LOAD DATA
# ------------------------------------------------------------
if not (os.path.exists(TRAIN_FILE) and os.path.exists(TEST_FILE)):
    from google.colab import files
    files.upload()

train = pd.read_csv(TRAIN_FILE)
test  = pd.read_csv(TEST_FILE)            # no labels -> only for final prediction
X, y  = train[FEATURES].values, train[TARGET].values
X_ext = test[FEATURES].values
print("Train file:", train.shape, "| Test file:", test.shape)

def rmse(a, b):
    return float(np.sqrt(np.mean((np.asarray(a) - np.asarray(b)) ** 2)))

def split_label(t):
    return f"{round((1 - t) * 100)}-{round(t * 100)}"

# ------------------------------------------------------------
# 2. PIPELINES + PARAMETER GRIDS
#    PolynomialFeatures + StandardScaler are cached, so they are computed once per
#    (degree, fold) and re-used for every alpha / l1_ratio / model.
# ------------------------------------------------------------
def make_models(memory):
    def pipe(reg):
        return Pipeline([("poly", PolynomialFeatures(include_bias=False)),
                         ("scale", StandardScaler()),
                         ("reg", reg)], memory=memory)
    return {
        "ridge": (pipe(Ridge()),
                  {"poly__degree": DEGREES, "reg__alpha": RIDGE_ALPHAS}),
        "lasso": (pipe(Lasso(max_iter=MAX_ITER, tol=TOL, random_state=SEED)),
                  {"poly__degree": DEGREES, "reg__alpha": LASSO_ALPHAS}),
        "elasticnet": (pipe(ElasticNet(max_iter=MAX_ITER, tol=TOL, random_state=SEED)),
                       {"poly__degree": DEGREES, "reg__alpha": ENET_ALPHAS, "reg__l1_ratio": ENET_L1}),
    }

cv = KFold(n_splits=N_FOLDS, shuffle=True, random_state=SEED)
scoring = {"rmse": "neg_root_mean_squared_error", "r2": "r2"}
ALPHA_RANGE = {"ridge": RIDGE_ALPHAS, "lasso": LASSO_ALPHAS, "elasticnet": ENET_ALPHAS}

# ------------------------------------------------------------
# 3. RUN ONE SPLIT:  GridSearchCV for each model, then score on that split's internal test
# ------------------------------------------------------------
def run_split(test_frac):
    label = split_label(test_frac)
    X_dev, X_te, y_dev, y_te = train_test_split(X, y, test_size=test_frac, random_state=SEED)
    print(f"\n{'=' * 70}\nSPLIT {label}:  dev (CV) = {len(X_dev)} rows | internal test = {len(X_te)} rows\n{'=' * 70}")

    cache_dir = tempfile.mkdtemp()
    models = make_models(Memory(location=cache_dir, verbose=0))
    searches, cv_tables, summary, preds = {}, [], [], {}
    for kind, (pipe, grid) in models.items():
        n_combo = int(np.prod([len(v) for v in grid.values()]))
        print(f"[{label}] {kind}: {n_combo} combinations x {N_FOLDS} folds = {n_combo * N_FOLDS} fits ...", flush=True)
        t0 = time.time()
        gs = GridSearchCV(pipe, grid, scoring=scoring, refit="rmse", cv=cv,
                          n_jobs=N_JOBS, return_train_score=True, error_score="raise")
        gs.fit(X_dev, y_dev)
        secs = time.time() - t0
        bp = gs.best_params_
        print(f"[{label}] {kind}: done in {secs:.0f}s | best params: "
              f"degree={bp['poly__degree']}, alpha={bp['reg__alpha']:.4g}"
              + (f", l1_ratio={bp['reg__l1_ratio']}" if 'reg__l1_ratio' in bp else "")
              + f" | CV RMSE {-gs.best_score_:.4f}", flush=True)
        # warn if the best value sits on the edge of a grid (a wider grid might do better)
        a_min, a_max = ALPHA_RANGE[kind].min(), ALPHA_RANGE[kind].max()
        if np.isclose(bp["reg__alpha"], a_min) or np.isclose(bp["reg__alpha"], a_max):
            print(f"      note: best alpha is at the edge of its grid ({a_min:.0e}..{a_max:.0e})")
        if bp["poly__degree"] in (min(DEGREES), max(DEGREES)):
            print("      note: best degree is at the edge of DEGREES")

        r = pd.DataFrame(gs.cv_results_)
        cv_tables.append(pd.DataFrame({
            "split": label, "model": kind,
            "degree": r["param_poly__degree"].astype(int),
            "alpha": r["param_reg__alpha"].astype(float),
            "l1_ratio": r["param_reg__l1_ratio"].astype(float) if "param_reg__l1_ratio" in r else np.nan,
            "train_rmse": -r["mean_train_rmse"], "cv_rmse": -r["mean_test_rmse"],
            "cv_rmse_std": r["std_test_rmse"],
            "train_r2": r["mean_train_r2"], "cv_r2": r["mean_test_r2"], "cv_r2_std": r["std_test_r2"],
            "rank": r["rank_test_rmse"]}))

        p = gs.best_estimator_.predict(X_te)        # already refit on the whole dev set
        preds[kind], searches[kind] = p, gs
        summary.append(dict(split=label, model=kind, best_degree=bp["poly__degree"],
                            best_alpha=bp["reg__alpha"], best_l1_ratio=bp.get("reg__l1_ratio", np.nan),
                            cv_rmse=-gs.best_score_, cv_r2=r["mean_test_r2"][gs.best_index_],
                            test_rmse=rmse(y_te, p), test_mae=mean_absolute_error(y_te, p),
                            test_r2=r2_score(y_te, p), seconds=secs))
    shutil.rmtree(cache_dir, ignore_errors=True)    # free the disk cache before the next split

    comp = pd.DataFrame(summary).set_index("model")
    base = rmse(y_te, np.full_like(y_te, y_dev.mean()))
    winner = comp["cv_rmse"].idxmin()               # chosen by CV only (no leakage)
    print(f"\n--- Split {label}: final comparison (scored on this split's internal test) ---")
    display(comp.drop(columns=["split"]).round(4))
    print(f"Baseline RMSE (always predict the mean): {base:.4f}  (R^2 = 0)")
    print(f">>> Winner for split {label}: {winner}  "
          f"(test RMSE {comp.loc[winner, 'test_rmse']:.4f}, R^2 {comp.loc[winner, 'test_r2']:.4f})")
    return dict(label=label, frac=test_frac, y_te=y_te, preds=preds, searches=searches,
                full=pd.concat(cv_tables, ignore_index=True), comp=comp, winner=winner, baseline=base)

T_START = time.time()
results = [run_split(t) for t in TEST_FRACTIONS]
print(f"\nAll splits finished in {(time.time() - T_START) / 60:.1f} minutes.")

# ------------------------------------------------------------
# 4. SAVE EVERYTHING + COMPARE THE SPLITS
# ------------------------------------------------------------
full_all = pd.concat([r["full"] for r in results], ignore_index=True)
best_per_degree_all = (full_all.loc[full_all.groupby(["split", "model", "degree"])["cv_rmse"].idxmin()]
                       .sort_values(["split", "model", "degree"]).reset_index(drop=True))
split_table = pd.concat([r["comp"].reset_index() for r in results], ignore_index=True)
full_all.sort_values(["split", "model", "rank"]).to_csv("gridsearch_full_results_var2.csv", index=False)
best_per_degree_all.to_csv("gridsearch_best_per_degree_var2.csv", index=False)
split_table.to_csv("split_comparison_var2.csv", index=False)

print("\n=== ALL SPLITS x ALL MODELS (best params, CV and internal-test scores) ===")
display(split_table.round(4))
for col, name in [("test_rmse", "Internal-test RMSE"), ("test_r2", "Internal-test R^2"), ("cv_rmse", "CV RMSE")]:
    print(f"\n{name}  (rows = split, columns = model):")
    display(split_table.pivot(index="split", columns="model", values=col)
            .reindex([r["label"] for r in results]).round(4))

# Best split = the split WINNER (picked by CV within the split) with the lowest internal-test RMSE.
win_rows = pd.DataFrame([dict(split=r["label"], model=r["winner"],
                              cv_rmse=r["comp"].loc[r["winner"], "cv_rmse"],
                              test_rmse=r["comp"].loc[r["winner"], "test_rmse"],
                              test_r2=r["comp"].loc[r["winner"], "test_r2"]) for r in results])
print("\n=== Winner of each split ===")
display(win_rows.round(4))
BEST = min(results, key=lambda r: r["comp"].loc[r["winner"], "test_rmse"])
BEST_KIND = BEST["winner"]
print(f"\n>>> BEST OVERALL: split {BEST['label']} | model {BEST_KIND} | "
      f"params {BEST['searches'][BEST_KIND].best_params_}")
print("    (Differences of a few percent between splits are partly random: each split has a")
print("     different internal-test set. Treat close results as ties.)")

# ------------------------------------------------------------
# 5. PLOTS ACROSS SPLITS
# ------------------------------------------------------------
labels = [r["label"] for r in results]
fig, axes = plt.subplots(1, 3, figsize=(17, 4.5))
w = 0.25
for i, kind in enumerate(KINDS):
    sub = split_table[split_table.model == kind].set_index("split").reindex(labels)
    for ax, col, ttl in zip(axes, ["test_rmse", "test_r2", "cv_rmse"],
                            ["Internal-test RMSE (lower = better)", "Internal-test R^2 (higher = better)",
                             "CV RMSE (lower = better)"]):
        ax.bar(np.arange(len(labels)) + (i - 1) * w, sub[col].values, w, label=kind)
        ax.set_title(ttl)
for ax in axes:
    ax.set_xticks(range(len(labels))); ax.set_xticklabels(labels); ax.set_xlabel("train-test split")
    ax.grid(alpha=.3, axis="y"); ax.legend()
axes[1].set_ylim(max(0, split_table["test_r2"].min() - 0.02), 1.0)
plt.tight_layout(); plt.show()

# CV RMSE / CV R^2 vs degree, one line per split, one panel per model
fig, axes = plt.subplots(2, 3, figsize=(18, 8))
for j, kind in enumerate(KINDS):
    for lab in labels:
        g = best_per_degree_all[(best_per_degree_all.model == kind) & (best_per_degree_all.split == lab)]
        axes[0, j].plot(g["degree"], g["cv_rmse"], marker="o", ms=3, label=lab)
        axes[1, j].plot(g["degree"], g["cv_r2"], marker="o", ms=3, label=lab)
    axes[0, j].set_title(f"{kind}: CV RMSE vs degree"); axes[0, j].set_yscale("log")
    axes[1, j].set_title(f"{kind}: CV R^2 vs degree")
    axes[1, j].set_ylim(max(0, best_per_degree_all["cv_r2"].min()), 1.005)
    for ax in axes[:, j]:
        ax.set_xlabel("polynomial degree"); ax.grid(alpha=.3); ax.legend(title="split", fontsize=8)
plt.tight_layout(); plt.show()

# ------------------------------------------------------------
# 6. FULL REPORT FOR THE BEST SPLIT
# ------------------------------------------------------------
B = BEST["label"]
full_b = BEST["full"]
bpd = best_per_degree_all[best_per_degree_all.split == B].reset_index(drop=True)

print(f"\n################  DETAILED REPORT: BEST SPLIT {B}  ################")
print("\n=== Top 10 combinations per model (by CV RMSE) ===")
for kind in KINDS:
    print(f"\n{kind}:")
    display(full_b[full_b.model == kind].sort_values("rank").head(10).round(4).reset_index(drop=True))

print("\n=== Best alpha / l1_ratio at each degree (train RMSE, CV RMSE, CV R^2 and their std) ===")
display(bpd.round(4))
for col, name in [("cv_r2", "CV R^2"), ("cv_rmse", "CV RMSE"), ("train_r2", "Train R^2")]:
    print(f"\n{name}  (rows = degree, columns = model):")
    display(bpd.pivot(index="degree", columns="model", values=col).round(4))

print(f"\n=== Final comparison for split {B} (internal test) ===")
display(BEST["comp"].drop(columns=["split"]).round(4))
print(f"Baseline RMSE (always predict the mean): {BEST['baseline']:.4f}  (R^2 = 0)")

# R^2 / RMSE vs degree
fig, axes = plt.subplots(1, 3, figsize=(17, 4.5))
for kind, g in bpd.groupby("model"):
    axes[0].plot(g["degree"], g["cv_r2"], marker="o", label=kind)
    axes[1].plot(g["degree"], g["train_r2"], marker="o", label=kind)
    axes[2].plot(g["degree"], g["cv_rmse"], marker="o", label=kind)
axes[0].set_title(f"CV R^2 (best alpha per degree) - split {B}")
axes[1].set_title("Train R^2 (best alpha per degree)")
axes[2].set_title("CV RMSE (best alpha per degree)"); axes[2].set_yscale("log")
axes[0].set_ylim(max(0, bpd["cv_r2"].min()), 1.005)
axes[1].set_ylim(max(0, bpd["train_r2"].min()), 1.005)
for ax in axes:
    ax.set_xlabel("polynomial degree"); ax.grid(alpha=.3); ax.legend()
plt.tight_layout(); plt.show()

# Heatmap of CV RMSE over degree x alpha (ElasticNet: best l1_ratio in each cell)
fig, axes = plt.subplots(1, 3, figsize=(18, 5.5))
for ax, kind in zip(axes, KINDS):
    g = full_b[full_b.model == kind]
    heat = g.groupby(["degree", "alpha"])["cv_rmse"].min().unstack("alpha")
    im = ax.imshow(np.log10(heat.values), aspect="auto", cmap="viridis_r")
    ax.set_xticks(range(len(heat.columns)))
    ax.set_xticklabels([f"{a:.0e}" for a in heat.columns], rotation=60)
    ax.set_yticks(range(len(heat.index))); ax.set_yticklabels(heat.index)
    ax.set_xlabel("alpha"); ax.set_ylabel("degree"); ax.set_title(f"{kind}: log10(CV RMSE), split {B}")
    fig.colorbar(im, ax=ax)
plt.tight_layout(); plt.show()

# Actual vs predicted on the internal test set
y_te_b = BEST["y_te"]
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, kind in zip(axes, KINDS):
    p = BEST["preds"][kind]; lim = [min(y_te_b.min(), p.min()), max(y_te_b.max(), p.max())]
    ax.scatter(y_te_b, p, s=10, alpha=.6); ax.plot(lim, lim, "r--")
    ax.set_title(f"{kind} (deg {int(BEST['comp'].loc[kind, 'best_degree'])})  "
                 f"R2 {BEST['comp'].loc[kind, 'test_r2']:.4f}")
    ax.set_xlabel("actual y"); ax.set_ylabel("predicted y")
plt.tight_layout(); plt.show()

# ------------------------------------------------------------
# 7. REFIT THE BEST MODEL ON ALL TRAINING ROWS, PREDICT THE EXTERNAL TEST FILE
# ------------------------------------------------------------
final_model = clone(BEST["searches"][BEST_KIND].best_estimator_).set_params(memory=None)
final_model.fit(X, y)
test_pred = final_model.predict(X_ext)

print(f"\nFinal model: {BEST_KIND} | {BEST['searches'][BEST_KIND].best_params_} | chosen on split {B}")
print("Test prediction range:", test_pred.min().round(2), "to", test_pred.max().round(2))
print("Train y range        :", y.min().round(2), "to", y.max().round(2))
span = y.max() - y.min()
if test_pred.min() < y.min() - 0.25 * span or test_pred.max() > y.max() + 0.25 * span:
    print("WARNING: predictions fall far outside the training y range - the polynomial may be blowing up.")

pd.DataFrame({"y": test_pred}).to_csv(OUT_FILE, index=False)
print("Saved", OUT_FILE, (len(test_pred), 1))

try:
    from google.colab import files
    for f in [OUT_FILE, "gridsearch_full_results_var2.csv",
              "gridsearch_best_per_degree_var2.csv", "split_comparison_var2.csv"]:
        files.download(f)
except ImportError:
    pass
```
